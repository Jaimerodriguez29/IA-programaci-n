# Prueba de profundidad en el MLP de Iris

Se compara la versión que viene en el ejemplo con otra que tiene dos capas ocultas adicionales. No se cambian épocas, tasa de aprendizaje, activación ni pérdida.


```python
import time
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from sklearn.datasets import load_iris
from tensorflow.keras.utils import to_categorical
```

```python
datos = load_iris()
entradas = datos.data
clases = datos.target
salidas = to_categorical(clases).astype('uint8')

print('Entradas:', entradas.shape)
print('Salidas:', salidas.shape)
```

### Red inicial


```python
red_inicial = keras.Sequential()
red_inicial.add(keras.Input(shape=(4,)))
red_inicial.add(layers.Dense(3, activation='sigmoid', name='dense_1'))
red_inicial.add(layers.Dense(3, activation='sigmoid', name='dense_2'))
red_inicial.summary()

red_inicial.compile(
    optimizer=tf.keras.optimizers.SGD(learning_rate=0.03),
    loss='mse'
)

t0 = time.perf_counter()
resultado_inicial = red_inicial.fit(entradas, salidas, epochs=500, verbose=0)
t_inicial = time.perf_counter() - t0

p1 = red_inicial.predict(entradas, verbose=0)
acc1 = (np.argmax(p1, axis=1) == clases).mean()
loss1 = resultado_inicial.history['loss'][-1]
print(f'loss={loss1:.6f} | exactitud={acc1*100:.2f}% | tiempo={t_inicial:.3f}s')
```

### Red con dos capas agregadas


```python
red_nueva = keras.Sequential()
red_nueva.add(keras.Input(shape=(4,)))
red_nueva.add(layers.Dense(3, activation='sigmoid', name='dense_1'))
red_nueva.add(layers.Dense(3, activation='sigmoid', name='dense_2'))
red_nueva.add(layers.Dense(3, activation='sigmoid', name='dense_3'))
red_nueva.add(layers.Dense(3, activation='sigmoid', name='dense_4'))
red_nueva.summary()

red_nueva.compile(
    optimizer=tf.keras.optimizers.SGD(learning_rate=0.03),
    loss='mse'
)

t0 = time.perf_counter()
resultado_nuevo = red_nueva.fit(entradas, salidas, epochs=500, verbose=0)
t_nuevo = time.perf_counter() - t0

p2 = red_nueva.predict(entradas, verbose=0)
acc2 = (np.argmax(p2, axis=1) == clases).mean()
loss2 = resultado_nuevo.history['loss'][-1]
print(f'loss={loss2:.6f} | exactitud={acc2*100:.2f}% | tiempo={t_nuevo:.3f}s')
```

```python
plt.figure(figsize=(8,4))
plt.plot(resultado_inicial.history['loss'], label='4x3x3')
plt.plot(resultado_nuevo.history['loss'], label='4x3x3x3x3')
plt.xlabel('Épocas')
plt.ylabel('MSE')
plt.title('Pérdida durante el entrenamiento')
plt.legend()
plt.grid(alpha=.2)
plt.show()

print('\nRESUMEN')
print(f'4x3x3: loss {loss1:.6f}, acc {acc1*100:.2f}%, {t_inicial:.3f}s')
print(f'4x3x3x3x3: loss {loss2:.6f}, acc {acc2*100:.2f}%, {t_nuevo:.3f}s')
```
